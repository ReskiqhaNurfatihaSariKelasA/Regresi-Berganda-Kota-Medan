# **REGRESI LINEAR BERGANDA – KOTA MEDAN**

Kota yang dipilih: Kota Medan

Variabel:

Variabel Dependent (Y) = KEMISKINAN

Variabel Independent (X₁) = PENGANGGURAN

Variabel Independent (X₂) = IPM

```python
# IMPORT LIBRARY
import numpy as np
import pandas as pd
import statsmodels.api as sm
import plotly.graph_objects as go
```

######IMPORT DATA

```python
# 1. LOAD DATA & FILTER 1 KOTA
df_all = pd.read_csv("https://raw.githubusercontent.com/Alam-A99/DATAAKSI/main/dataset_fiskal_97_hipoFull.csv")
df = df_all[df_all["Kota"] == "Kota Medan"].reset_index(drop=True)

# 2. PILIH VARIABEL & RAPIKAN NAMA
df = df[["Tahun", "KEMISKINAN", "PENGANGGURAN", "IPM"]].copy()
df.columns = ["Tahun", "Kemiskinan", "Pengangguran", "IPM"]
df = df.dropna().reset_index(drop=True)

print(df.shape)
df
```

######REGRESI BERGANDA

```python
# Variabel independen (X) dan dependen (Y)
X = df[["Pengangguran", "IPM"]]
y = df["Kemiskinan"]

# Menambahkan intercept
X_model = sm.add_constant(X)

# Estimasi OLS
model = sm.OLS(y, X_model)
results = model.fit()

print("=" * 70)
print(" HASIL REGRESI LINEAR BERGANDA - KOTA MEDAN")
print("=" * 70)
print(results.summary())

# Ambil koefisien
intercept = results.params["const"]
b1 = results.params["Pengangguran"]
b2 = results.params["IPM"]
r_squared = results.rsquared
adjusted_r_squared = results.rsquared_adj

# Persamaan regresi
persamaan = (
    f"Ŷ = {intercept:.3f}"
    f" + ({b1:.3f}) X1"
    f" + ({b2:.3f}) X2"
)

print("\n" + "=" * 70)
print(" PERSAMAAN REGRESI")
print("=" * 70)
print(persamaan)
print(f"\nR2 = {r_squared:.4f}")
print(f"Adjusted R2 = {adjusted_r_squared:.4f}")

# Nilai prediksi
df["Prediksi"] = results.predict(X_model)
print("\n" + "=" * 70)
print(" DATA AKTUAL & PREDIKSI")
print("=" * 70)
print(df.round(2))
```

######VISUALISASI 3D

```python
# Grid untuk bidang regresi (X1 dan X2)
x1 = np.linspace(df["Pengangguran"].min(), df["Pengangguran"].max(), 40)
x2 = np.linspace(df["IPM"].min(), df["IPM"].max(), 40)
X1, X2 = np.meshgrid(x1, x2)

# Nilai Y pada bidang regresi
Y = intercept + b1 * X1 + b2 * X2

fig = go.Figure()

# Data aktual
fig.add_trace(
    go.Scatter3d(
        x=df["Pengangguran"],
        y=df["IPM"],
        z=df["Kemiskinan"],
        mode="markers",
        marker=dict(size=7, color="red", opacity=0.9),
        name="Data Aktual",
        text=[f"Tahun {t}" for t in df["Tahun"]],
        hovertemplate=(
            "<b>%{text}</b><br>"
            "Pengangguran (X1): %{x}<br>"
            "IPM (X2): %{y}<br>"
            "Kemiskinan (Y): %{z}"
            "<extra></extra>"
        )
    )
)

# Bidang regresi
fig.add_trace(
    go.Surface(
        x=X1, y=X2, z=Y,
        opacity=0.60,
        colorscale="Viridis",
        name="Bidang Regresi",
        hovertemplate=(
            "Pengangguran (X1): %{x:.2f}<br>"
            "IPM (X2): %{y:.2f}<br>"
            "Prediksi Kemiskinan: %{z:.2f}"
            "<extra></extra>"
        )
    )
)

fig.update_layout(
    title={
        "text": (
            "REGRESI LINEAR BERGANDA 3D - KOTA MEDAN"
            "<br>"
            f"<sup>{persamaan} | R2 = {r_squared:.3f}</sup>"
        ), "x": 0.5
    },
    scene={
        "xaxis": {"title": "Pengangguran (X1)"},
        "yaxis": {"title": "IPM (X2)"},
        "zaxis": {"title": "Kemiskinan (Y)"}
    },
    width=1000,
    height=750,
    margin={"l": 0, "r": 0, "b": 0, "t": 120},
    legend={"x": 0.02, "y": 0.98}
)

fig.show()
fig.write_html("visualisasi_regresi_3d.html")
```

## INTERPRETASI HASIL REGRESI LINEAR BERGANDA – KOTA MEDAN

### 1. Interpretasi Persamaan Regresi

Berdasarkan hasil analisis regresi linear berganda menggunakan metode *Ordinary Least Squares* (OLS), diperoleh persamaan regresi sebagai berikut:

**Ŷ = -6,050 + 0,453 X₁ + 0,191 X₂**

Dalam persamaan tersebut, Y merupakan tingkat kemiskinan, X₁ merupakan tingkat pengangguran, dan X₂ merupakan Indeks Pembangunan Manusia (IPM).

Nilai konstanta sebesar -6,050 menunjukkan bahwa ketika tingkat pengangguran dan IPM bernilai nol, nilai kemiskinan yang diprediksi oleh model adalah -6,050. Namun, nilai ini lebih tepat dipahami sebagai konstanta matematis dalam model karena kondisi kedua variabel independen bernilai nol tidak sesuai dengan kondisi data yang diamati.

Koefisien pengangguran sebesar 0,453 menunjukkan bahwa setiap kenaikan satu satuan tingkat pengangguran diperkirakan berkaitan dengan kenaikan kemiskinan sebesar 0,453 satuan, dengan IPM dianggap tetap. Sementara itu, koefisien IPM sebesar 0,191 menunjukkan bahwa setiap kenaikan satu satuan IPM diperkirakan berkaitan dengan kenaikan kemiskinan sebesar 0,191 satuan, dengan tingkat pengangguran dianggap tetap. Kedua hubungan tersebut merupakan hasil estimasi model dan belum terbukti signifikan secara statistik.

### 2. Interpretasi Uji Parsial (Uji t)

Uji t digunakan untuk mengetahui apakah masing-masing variabel independen memiliki hubungan yang signifikan secara statistik dengan variabel dependen ketika variabel independen lainnya dikendalikan.

**a. Pengaruh pengangguran terhadap kemiskinan**

Variabel pengangguran memiliki koefisien regresi sebesar 0,453 dengan nilai probabilitas (*p-value*) sebesar 0,476. Karena nilai probabilitas lebih besar dari 0,05, pengangguran tidak menunjukkan hubungan yang signifikan secara statistik dengan kemiskinan pada taraf signifikansi 5%. Meskipun koefisiennya positif, hasil ini belum memberikan bukti yang cukup untuk menyimpulkan adanya hubungan yang signifikan antara pengangguran dan kemiskinan di Kota Medan selama periode pengamatan.

**b. Pengaruh IPM terhadap kemiskinan**

Variabel IPM memiliki koefisien regresi sebesar 0,191 dengan nilai probabilitas sebesar 0,452. Karena nilai probabilitas lebih besar dari 0,05, IPM juga tidak menunjukkan hubungan yang signifikan secara statistik dengan kemiskinan pada taraf signifikansi 5%. Dengan demikian, berdasarkan model ini, belum terdapat bukti yang cukup untuk menyatakan bahwa IPM memiliki hubungan yang signifikan dengan kemiskinan setelah tingkat pengangguran dikendalikan.

### 3. Interpretasi Uji Simultan (Uji F)

Berdasarkan hasil regresi, diperoleh nilai F-statistic sebesar 0,4893 dengan nilai probabilitas sebesar 0,630. Karena nilai probabilitas tersebut lebih besar dari 0,05, pengangguran dan IPM secara bersama-sama tidak menunjukkan pengaruh yang signifikan secara statistik terhadap kemiskinan dalam model regresi ini.

Artinya, pada taraf signifikansi 5%, model belum mampu menunjukkan hubungan simultan yang signifikan antara kedua variabel independen dan kemiskinan di Kota Medan selama periode 2014–2024.

### 4. Interpretasi Koefisien Determinasi (R²)

Nilai koefisien determinasi (R²) yang diperoleh adalah 0,109 atau 10,9%. Hal ini menunjukkan bahwa pengangguran dan IPM secara bersama-sama dapat menjelaskan sekitar 10,9% variasi kemiskinan dalam data yang dianalisis. Sementara itu, sekitar 89,1% variasi lainnya berkaitan dengan faktor-faktor di luar model serta variasi yang belum dapat dijelaskan oleh model regresi ini.

Nilai *Adjusted R²* sebesar -0,1137 menunjukkan bahwa setelah memperhitungkan jumlah variabel independen dan ukuran sampel, kemampuan model dalam menjelaskan variasi kemiskinan tergolong lemah. Nilai negatif ini tidak berarti bahwa kemiskinan bernilai negatif, tetapi menunjukkan bahwa model memiliki kemampuan penjelasan yang rendah dibandingkan model sederhana yang hanya menggunakan rata-rata variabel dependen.

### 5. Kesimpulan

Berdasarkan hasil regresi linear berganda terhadap data Kota Medan tahun 2014–2024, variabel pengangguran dan IPM memiliki koefisien regresi positif, tetapi keduanya tidak menunjukkan hubungan yang signifikan secara statistik dengan kemiskinan pada taraf signifikansi 5%. Secara simultan, kedua variabel tersebut juga tidak menunjukkan hubungan yang signifikan dengan kemiskinan.

Nilai R² sebesar 10,9% menunjukkan bahwa kemampuan model dalam menjelaskan variasi kemiskinan masih terbatas. Oleh karena itu, hasil analisis ini belum memberikan bukti statistik yang cukup untuk menyatakan bahwa pengangguran dan IPM berpengaruh signifikan terhadap kemiskinan di Kota Medan. Penelitian selanjutnya dapat mempertimbangkan penambahan variabel lain yang relevan dan penggunaan data dengan jumlah observasi lebih besar agar analisis menjadi lebih informatif.
